```python
import numpy as np
def softmax(Z: np.ndarray) -> np.ndarray:
    # Z : a numpy ndarray of shape (N, C)
    # each row of Z is the set of scores Z[i]
    # return a numpy ndarray of shape (N, C)
    max_Z = Z.max(axis=1, keepdims=True)
    e_Z = np.exp(Z - max_Z)
    return e_Z / e_Z.sum(axis=1, keepdims=True)
```

```python
def softmax_loss(W: np.ndarray, X: np.ndarray, Y: np.ndarray) -> np.ndarray:
    # W : a numpy ndarray of shape (d, C). Each column is the vector w_i for class C_i
    # X : a numpy ndarray of shape (N, d).  Each row corresponds to one data point
    # Y : a numpy ndarray of shape (N,). Each element is the class of each corresponding data point
    A = softmax(X @ W) # softmax for dataset
    cross_entropy = np.mean(np.log(A[range(Y.shape[0]), Y])) #  in this scope, Y[i] = j is a number like 0, 1, 2, 3, 4, 5, ...
    return cross_entropy
```
