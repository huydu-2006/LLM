## Model 
weight = w1 * height + w0

```python
import numpy as np
import matplotlib.pyplot as plt
import sklearn

# Design matrix X: shape (N, d) = (13, 1)  -- quy uoc sklearn
# Moi HANG la mot mau du lieu, moi COT la mot feature
X = np.array([[147],
              [150],
              [153],
              [158],
              [163],
              [165],
              [168],
              [170],
              [173],
              [175],
              [178],
              [180],
              [183]])

# Target vector y: shape (N,) -- Y[i] la nhan cua mau X[i]
Y = np.array([49, 50, 51, 54, 58, 59, 60, 62, 63, 64, 66, 67, 68])

print('X.shape =', X.shape, ' Y.shape =', Y.shape)
```

```python
# Building Xbar: them cot bias x0 = 1 vao ben trai -> (N, d+1) = (13, 2)
x0 = np.ones((X.shape[0], 1), dtype = X.dtype)
Xbar = np.concat((x0, X), axis = 1)

# Calculating weights of the linear regression model
# w = (Xbar^T Xbar)^-1 Xbar^T y
A = Xbar.T @ Xbar
b = Xbar.T @ Y
W = np.linalg.pinv(A) @ b

# Du doan: y_hat = Xbar @ w
Yhead = Xbar @ W

print('Xbar.shape =', Xbar.shape)
print('A.shape =', A.shape, ' b.shape =', b.shape, ' W.shape =', W.shape)
print('Yhead.shape =', Yhead.shape)
print('w0 =', W[0], ' w1 =', W[1])
```

```python
# Plotting dataset
plt.scatter(X, Y, label='Dataset');
plt.plot(X, Yhead, color='red', label='Model result');
plt.legend();
```

```python
# Checking with answer from Scikitlearn library
# X da dung quy uoc (n_samples, n_features) -> goi truc tiep, khong can chuyen vi
regr = sklearn.linear_model.LinearRegression()
regr.fit(X, Y)
print('w1 =', regr.coef_[0])
print('w0 =', regr.intercept_)
print()
print('Khop voi closed-form:', np.allclose(W[1], regr.coef_[0]), np.allclose(W[0], regr.intercept_))
```
